```python
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl

rating = ctrl.Antecedent(np.arange(0, 5.1, 0.1), 'store_rating')
rating['Thap']       = fuzz.trimf(rating.universe, [0, 0, 4.0])
rating['Trung_binh'] = fuzz.trimf(rating.universe, [3.8, 4.25, 4.6])
rating['Cao']        = fuzz.trimf(rating.universe, [4.4, 5.0, 5.0])

sales = ctrl.Antecedent(np.arange(0, 1001, 10), 'sales_volume')
sales['Thap']       = fuzz.trimf(sales.universe, [0, 0, 150])
sales['Trung_binh'] = fuzz.trimf(sales.universe, [100, 300, 600])
sales['Cao']        = fuzz.trimf(sales.universe, [500, 800, 1000])

profit = ctrl.Antecedent(np.arange(0, 51, 1), 'profit_margin')
profit['Thap']       = fuzz.trimf(profit.universe, [0, 0, 15])
profit['Trung_binh'] = fuzz.trimf(profit.universe, [10, 20, 30])
profit['Cao']        = fuzz.trimf(profit.universe, [25, 40, 50])

season = ctrl.Antecedent(np.arange(0, 11, 1), 'seasonal_event')
season['Khong_co']   = fuzz.trimf(season.universe, [0, 0, 3])
season['Trung_binh'] = fuzz.trimf(season.universe, [2, 5, 7])
season['Cao']        = fuzz.trimf(season.universe, [6, 9, 10])

competitor = ctrl.Antecedent(np.arange(0, 71, 1), 'competitor_discount')
competitor['Thap']       = fuzz.trimf(competitor.universe, [0, 0, 15])
competitor['Trung_binh'] = fuzz.trimf(competitor.universe, [10, 25, 40])
competitor['Cao']        = fuzz.trimf(competitor.universe, [35, 55, 70])

discount = ctrl.Consequent(np.arange(0, 71, 1), 'discount_percentage')
discount['Rat_thap']  = fuzz.trimf(discount.universe, [0, 0, 7])
discount['Thap']      = fuzz.trimf(discount.universe, [3, 7, 12])
discount['Trung_binh']= fuzz.trimf(discount.universe, [10, 15, 25])
discount['Cao']       = fuzz.trimf(discount.universe, [20, 30, 45])
discount['Rat_cao']   = fuzz.trimf(discount.universe, [40, 55, 70])

rules = []
rules.append(ctrl.Rule(rating['Cao'] & sales['Cao'] & profit['Cao'], discount['Rat_thap']))
rules.append(ctrl.Rule(rating['Thap'] & sales['Thap'] & profit['Cao'], discount['Cao']))
rules.append(ctrl.Rule(season['Cao'] & competitor['Cao'], discount['Rat_cao']))
rules.append(ctrl.Rule(rating['Trung_binh'] & sales['Trung_binh'] & profit['Trung_binh'], discount['Trung_binh']))
rules.append(ctrl.Rule(competitor['Thap'] & sales['Cao'], discount['Rat_thap']))
rules.append(ctrl.Rule(rating['Thap'] & season['Khong_co'], discount['Trung_binh']))
rules.append(ctrl.Rule(sales['Thap'] & profit['Thap'], discount['Rat_cao']))

discount_ctrl = ctrl.ControlSystem(rules)
discount_sim = ctrl.ControlSystemSimulation(discount_ctrl)

def predict_discount(store_rating, sales_volume, profit_margin, seasonal_level, competitor_disc):
    discount_sim.input['store_rating'] = store_rating
    discount_sim.input['sales_volume'] = sales_volume
    discount_sim.input['profit_margin'] = profit_margin
    discount_sim.input['seasonal_event'] = seasonal_level
    discount_sim.input['competitor_discount'] = competitor_disc

    discount_sim.compute()
    return round(discount_sim.output['discount_percentage'], 1)

if __name__ == "__main__":
    print("MÔ HÌNH FUZZY LOGIC QUYẾT ĐỊNH GIẢM GIÁ SHOPEE \n")

    store_rating = 4.3
    sales_volume = 400
    profit_margin = 12
    seasonal_level = 8
    competitor_disc = 45

    ket_qua = predict_discount(store_rating, sales_volume, profit_margin, seasonal_level, competitor_disc)

    print("Trường hợp phân tích theo đề bài:")
    print(f"- Xếp hạng cửa hàng     : {store_rating} → Trung bình")
    print(f"- Khối lượng bán hàng   : Trung bình")
    print(f"- Biên lợi nhuận        : Thấp ({profit_margin}%)")
    print(f"- Sự kiện theo mùa      : Cao")
    print(f"- Giảm giá đối thủ      : Cao ({competitor_disc}%)")
    print("-" * 60)
    print(f"→ Kết quả chiết khấu đề xuất: **{ket_qua}%** (Mức Cao: 30-40%)")
```
